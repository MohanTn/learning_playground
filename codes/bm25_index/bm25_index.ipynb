```python
import re

from rank_bm25 import BM25Okapi
```

```python
def tokenize(text: str) -> list[str]:
    return re.findall(r"[a-z0-9]+", text.lower())
```

```python
# Sample corpus: one string per document
documents = [
    "The quick brown fox jumps over the lazy dog",
    "BM25 is a ranking function used by search engines to estimate relevance",
    "Python is a popular programming language for data science and scripting",
    "The lazy dog slept in the sun all afternoon",
    "Search engines rank documents using term frequency and inverse document frequency",
    "Foxes are wild animals related to dogs and wolves",
]

tokenized_corpus = [tokenize(doc) for doc in documents]
```

```python
bm25 = BM25Okapi(tokenized_corpus)
```

```python
def search(query: str, top_k: int = 3) -> list[tuple[float, str]]:
    scores = bm25.get_scores(tokenize(query))
    ranked = sorted(zip(scores, documents), key=lambda pair: pair[0], reverse=True)
    return ranked[:top_k]
```

```python
query = "lazy dog search"
for score, doc in search(query):
    print(f"{score:.4f}  {doc}")
```
